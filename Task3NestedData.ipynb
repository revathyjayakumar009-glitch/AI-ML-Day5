```python
#Create list of 5 students
students = [
    {"name": "Jithu", "age": 20, "marks": 85},
    {"name": "Rahul", "age": 21, "marks": 90},
    {"name": "Joshna", "age": 19, "marks": 88},
    {"name": "Arjun", "age": 22, "marks": 76},
    {"name": "Aswin", "age": 20, "marks": 92}
]
```

```python
# print all students
for i in students:
 print(i)
```

```python
#Find Average marks
total =0
for i in students:
    total+=i["marks"]

average=total/len(students)
print("Average Marks:",average)
    
```

```python
#highest marks
highest_marks = max(students, key=lambda x: x["marks"])

print("Student_Name:", highest_marks["name"])
print("Marks:", highest_marks["marks"])
```

```python
#print students with marks>75
for i in students:
    if i["marks"]>75:
        print(i)
```
