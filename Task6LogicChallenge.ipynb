```python
#Second Highest number in list
numbers = [10, 20, 4, 99, 45]
numbers.sort()
print("Second Highest Number in the List:",numbers[-2])
```

```python
#Count frequency of characters
text = input("Enter a string: ")

frequency = {}

for char in text:
    if char in frequency:
        frequency[char] += 1
    else:
        frequency[char] = 1

print(frequency)
```
